```python
import pandas as pd
train = pd.read_csv("train.csv")
test = pd.read_csv("test.csv")
```

```python
print(train.describe())
```

```python
print(train.dtypes)
print("\nMissing values:")
print(train.isna().sum()[train.isna().sum() > 0])

print("\nDuplicate rows:", train.duplicated().sum())
print("Duplicate IDs:", train["id"].duplicated().sum())

print("\nTarget:")
print(train["satisfaction"].value_counts(dropna=False))
print(train["satisfaction"].value_counts(normalize=True, dropna=False))
```

```python
print("\nCategorical columns:")
print(train.select_dtypes(include=["object", "category"]).columns.tolist())

print("\nNumerical columns:")
print(train.select_dtypes(include=["number"]).columns.tolist())
```

```python
# Train/test distribution check
for col in train.columns:
    if col == "satisfaction":
        continue

    if train[col].dtype == "object":
        print(f"\n{col}")
        print("TRAIN:", train[col].value_counts(normalize=True).head(10))
        print("TEST :", test[col].value_counts(normalize=True).head(10))
```

```python
import pandas as pd

num_cols = train.select_dtypes(include="number").columns.drop("id")

distribution_check = pd.DataFrame({
    "train_mean": train[num_cols].mean(),
    "test_mean": test[num_cols].mean(),
    "train_std": train[num_cols].std(),
    "test_std": test[num_cols].std(),
})

distribution_check["mean_diff"] = (
    distribution_check["test_mean"] -
    distribution_check["train_mean"]
)

distribution_check.sort_values("mean_diff", key=abs, ascending=False)
```
