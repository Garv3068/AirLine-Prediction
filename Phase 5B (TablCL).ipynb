```python
import gc
import numpy as np
import pandas as pd
import torch

from sklearn.metrics import roc_auc_score
from tabicl import TabICLClassifier

# ============================================================
# TABICL — MEMORY-SAFE 10K BENCHMARK
# ============================================================

print("CUDA:", torch.cuda.is_available())
print("GPU:", torch.cuda.get_device_name(0))

# Use the existing train dataframe.
# DO NOT create another full copy.
feature_cols = [
    c for c in train.columns
    if c not in ["id", "satisfaction"]
]

categorical_cols = [
    "Gender",
    "Customer Type",
    "Type of Travel",
    "Class"
]

# 10K stratified sample
from sklearn.model_selection import train_test_split

sample_idx, _ = train_test_split(
    np.arange(len(train)),
    train_size=10000,
    stratify=train["satisfaction"],
    random_state=42
)

tabicl_data = train.iloc[sample_idx]

X_tabicl = tabicl_data[feature_cols].copy()
y_tabicl = tabicl_data["satisfaction"].astype(np.int8).to_numpy()

# Let TabICL handle categorical columns
for col in categorical_cols:
    X_tabicl[col] = X_tabicl[col].astype("category")

del tabicl_data, sample_idx
gc.collect()

print("TabICL training shape:", X_tabicl.shape)
print("Target:", y_tabicl.shape)

# ------------------------------------------------------------
# Model
# ------------------------------------------------------------

tabicl_model = TabICLClassifier(
    n_estimators=4,
    batch_size=2,
    kv_cache=False,
    offload_mode="auto",
    use_amp=True,
    device="cuda",
    random_state=42,
    verbose=True
)

print("\nFitting TabICL...")

tabicl_model.fit(
    X_tabicl,
    y_tabicl
)

print("Fit complete.")
```

```python
# ============================================================
# TABICL — VALIDATION PREDICTION
# ============================================================

print("Validation shape:", X_valid.shape)

tabicl_valid_pred = tabicl_model.predict_proba(
    X_valid
)[:, 1]

print("Prediction shape:", tabicl_valid_pred.shape)

tabicl_auc = roc_auc_score(
    y_valid,
    tabicl_valid_pred
)

print(f"\nTabICL 10K AUC: {tabicl_auc:.9f}")
print(f"Current Champion: 0.958378205")
print(f"Gap: {tabicl_auc - 0.958378205:+.9f}")
```

```python
from sklearn.metrics import roc_auc_score
import numpy as np

# ------------------------------------------------------------
# Reconstruct current champion
# ------------------------------------------------------------

lgb_v  = np.load("/kaggle/input/datasets/garv3068/dataset/artifacts/oof/lgb_optimized.npy")
xgb_v  = np.load("/kaggle/input/datasets/garv3068/dataset/artifacts/oof/baseline_xgb.npy")
cat_v  = np.load("/kaggle/input/datasets/garv3068/dataset/artifacts/oof/baseline_cat.npy")
spec_v = np.load("/kaggle/input/datasets/garv3068/dataset/artifacts/oof/lgb_specialized_service.npy")
div_v  = np.load("/kaggle/input/datasets/garv3068/dataset/artifacts/oof/exp_diversity_lgb01_blend.npy")

old_champion_valid = (
    0.57 * lgb_v
    + 0.24 * xgb_v
    + 0.19 * cat_v
)

champion_valid = (
    0.90 * (
        0.75 * old_champion_valid
        + 0.25 * spec_v
    )
    + 0.10 * (
        div_v
    )
)

# ------------------------------------------------------------
# Correlation
# ------------------------------------------------------------

corr = np.corrcoef(
    champion_valid,
    tabicl_valid_pred
)[0, 1]

print(f"Champion ↔ TabICL correlation: {corr:.9f}")

# ------------------------------------------------------------
# Tiny blend search
# ------------------------------------------------------------

print("\nTabICL blend search")
print("=" * 50)

best_auc = -1
best_weight = None

for w in np.arange(0.00, 0.101, 0.01):

    blend = (
        (1 - w) * champion_valid
        + w * tabicl_valid_pred
    )

    auc = roc_auc_score(y_valid, blend)

    print(
        f"TabICL={w:.2f} | "
        f"Champion={1-w:.2f} | "
        f"AUC={auc:.9f}"
    )

    if auc > best_auc:
        best_auc = auc
        best_weight = w

print("\nBEST")
print(f"Weight: {best_weight:.2f}")
print(f"AUC:    {best_auc:.9f}")
print(f"Gain:   {best_auc - 0.958378205:+.9f}")
```

```python
# ============================================================
# TABICL 50K + 100K COMBINATION
# ============================================================

from sklearn.metrics import roc_auc_score

best_auc = -1
best_w50 = None
best_w100 = None

print("TabICL 50K + 100K ensemble search")
print("=" * 65)

# Total TabICL weight from 8% to 16%
# Split between 50K and 100K.

for total in [0.08, 0.10, 0.12, 0.14, 0.16]:

    for w50_fraction in [
        0.00,
        0.25,
        0.50,
        0.75,
        1.00
    ]:

        w50 = total * w50_fraction
        w100 = total - w50

        blend = (
            (1 - total) * champion_valid
            + w50 * tabicl_50k_valid_pred
            + w100 * tabicl_100k_valid_pred
        )

        auc = roc_auc_score(
            y_valid,
            blend
        )

        print(
            f"Total={total:.2f} | "
            f"50K={w50:.3f} | "
            f"100K={w100:.3f} | "
            f"AUC={auc:.9f}"
        )

        if auc > best_auc:
            best_auc = auc
            best_w50 = w50
            best_w100 = w100

print("\n" + "=" * 65)
print("BEST COMBINATION")
print(f"Champion: {1-best_w50-best_w100:.3f}")
print(f"TabICL 50K: {best_w50:.3f}")
print(f"TabICL 100K: {best_w100:.3f}")
print(f"AUC: {best_auc:.9f}")
print(f"Gain vs 0.958378205: {best_auc - 0.958378205:+.9f}")
```
