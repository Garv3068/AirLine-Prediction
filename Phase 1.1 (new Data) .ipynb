```python
import pandas as pd
import numpy as np

# Load
source = pd.read_csv("new Dataset/train.csv")
comp = pd.read_csv("train.csv")

# Remove irrelevant columns
source = source.drop(columns=["Unnamed: 0"], errors="ignore")

print("SOURCE:", source.shape)
print("COMPETITION:", comp.shape)

print("\nSOURCE COLUMNS:")
print(source.columns.tolist())

print("\nCOMPETITION COLUMNS:")
print(comp.columns.tolist())
```

```python
# ------------------------------------------------------------
# 1. Column comparison
# ------------------------------------------------------------

source_features = set(source.columns) - {"satisfaction", "id"}
comp_features = set(comp.columns) - {"satisfaction", "id"}

print("\nOnly in SOURCE:")
print(sorted(source_features - comp_features))

print("\nOnly in COMPETITION:")
print(sorted(comp_features - source_features))

print("\nCommon features:")
print(len(source_features & comp_features))
```

```python
numeric_cols = [
    c for c in source_features
    if c in comp.columns and
    pd.api.types.is_numeric_dtype(source[c])
]

comparison = []

for col in numeric_cols:
    comparison.append({
        "feature": col,

        "source_mean": source[col].mean(),
        "competition_mean": comp[col].mean(),

        "source_std": source[col].std(),
        "competition_std": comp[col].std(),

        "source_median": source[col].median(),
        "competition_median": comp[col].median(),

        "mean_diff_%": (
            (comp[col].mean() - source[col].mean())
            / source[col].mean() * 100
            if source[col].mean() != 0 else np.nan
        )
    })

dist_df = pd.DataFrame(comparison)

pd.set_option("display.max_columns", None)
print(dist_df.round(4))
```

```python
categorical_cols = [
    c for c in source_features
    if c in comp.columns and
    source[c].dtype == "object"
]

for col in categorical_cols:

    print("\n" + "="*70)
    print(col)

    s = source[col].value_counts(normalize=True)
    c = comp[col].value_counts(normalize=True)

    comparison = pd.concat(
        [s.rename("source"),
         c.rename("competition")],
        axis=1
    ).fillna(0)

    comparison["difference"] = (
        comparison["competition"] -
        comparison["source"]
    )

    print(comparison.round(4))
```

```python
# ============================================================
# SOURCE vs COMPETITION — TARGET RELATIONSHIP ANALYSIS
# ============================================================

source["target"] = (
    source["satisfaction"] == "satisfied"
).astype(int)

comp["target"] = (
    comp["satisfaction"] == True
).astype(int)

common_features = [
    c for c in source.columns
    if c in comp.columns
    and c not in ["id", "satisfaction", "target"]
]

for col in common_features:

    print("\n" + "=" * 80)
    print(f"FEATURE: {col}")

    s_rate = (
        source.groupby(col)["target"]
        .agg(["mean", "count"])
        .rename(columns={
            "mean": "source_rate",
            "count": "source_count"
        })
    )

    c_rate = (
        comp.groupby(col)["target"]
        .agg(["mean", "count"])
        .rename(columns={
            "mean": "competition_rate",
            "count": "competition_count"
        })
    )

    result = s_rate.join(c_rate, how="outer")

    print(result.round(4))
```

```python
relationship_results = []

for col in common_features:

    s = source.groupby(col)["target"].mean()
    c = comp.groupby(col)["target"].mean()

    aligned = pd.concat(
        [s.rename("source"),
         c.rename("competition")],
        axis=1
    ).dropna()

    if len(aligned) >= 3:

        corr = aligned["source"].corr(
            aligned["competition"]
        )

        relationship_results.append({
            "feature": col,
            "relationship_corr": corr,
            "levels": len(aligned)
        })

relationship_df = (
    pd.DataFrame(relationship_results)
    .sort_values("relationship_corr", ascending=False)
)

print(relationship_df)
```
